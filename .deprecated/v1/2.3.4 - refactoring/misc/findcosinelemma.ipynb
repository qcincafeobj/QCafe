````python
import math
import random

def real_factory(seed: float, s: float) -> list[float]:
    # random.seed(seed)
    epsilon = round(math.pi / 100, ndigits=10)
    theta_s = s

    real_nums = []
    for _ in range(20):
        real = random.uniform(epsilon, theta_s / 3 - epsilon)
        real = round(real, ndigits=10)
        real_nums.append(real)
        theta_s = theta_s - real

    real_nums.append(theta_s)

    return real_nums

s = math.pi / 2
real_nums_1 = real_factory(seed=1.22325, s=s)
real_nums_2 = real_factory(seed=1.98715, s=s)

print(real_nums_1)
print(real_nums_2)
print(sum(real_nums_1) - s)
print(sum(real_nums_2) - s)
````
